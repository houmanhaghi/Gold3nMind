# **Arbitrage**
 is the simultaneous purchase and sale of an asset in different markets to exploit tiny differences in their prices. Arbitrage trades are most commonly made in stocks, commodities, and currencies, but can be accomplished in with any asset. Arbitrage takes advantage of the inevitable inefficiencies in markets.<br><br>

**Target website to gain information**


> https://arzex.io/arbitrage-list/

now we want to obtain this info:


1.   what is the best coin?
2.   which exchange is the best?

with respect to we would like to have this with the least number of coins and exchanges.

````python
# prompt: gain information about currency and exchanges in every minutes from this site: "https://arzex.io/arbitrage-list/"

import requests
from bs4 import BeautifulSoup
import time

def get_arbitrage_info(url):
    """
    Scrapes arbitrage information from the given URL.

    Args:
        url: The URL of the arbitrage list page.

    Returns:
        A dictionary containing the best coin and exchange based on the scraped data.
        Returns an empty dictionary if scraping fails or if no data is found.
    """

    try:
        response = requests.get(url)
        response.raise_for_status()  # Raise an exception for bad status codes

        soup = BeautifulSoup(response.content, "html.parser")

        coin_elements = soup.select(".coin-name")  # Replace with the actual CSS selector
        exchange_elements = soup.select(".exchange-name") # Replace with the actual CSS selector

        if not coin_elements or not exchange_elements:
          print("Could not find coin or exchange elements")
          return {}

        best_coin = coin_elements[0].text.strip() if coin_elements else "N/A"
        best_exchange = exchange_elements[0].text.strip() if exchange_elements else "N/A"

        return {"best_coin": best_coin, "best_exchange": best_exchange}

    except requests.exceptions.RequestException as e:
        print(f"Error fetching URL: {e}")
        return {}
    except Exception as e:
        print(f"An error occurred: {e}")
        return {}

if __name__ == "__main__":
    url = "https://arzex.io/arbitrage-list/"

    while True:
        arbitrage_data = get_arbitrage_info(url)
        if arbitrage_data:
            print(f"Best Coin: {arbitrage_data['best_coin']}")
            print(f"Best Exchange: {arbitrage_data['best_exchange']}")
        else:
            print("Failed to retrieve arbitrage information.")

        time.sleep(60)
````
